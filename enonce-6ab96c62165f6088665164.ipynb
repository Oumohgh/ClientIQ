
# 🧩 Brief Projet — Segmentation des clients avec K-Means et Random Forest


# 🎯 Objectif

L'objectif de ce projet est de construire une solution complète de **segmentation des clients** à partir de leurs comportements d'achat.

Dans un premier temps, une approche de **clustering non supervisé** sera utilisée afin d'identifier différents profils de clients.

Dans un second temps, les segments obtenus seront utilisés comme classes cibles afin de construire un modèle de **classification supervisée** capable d'affecter un nouveau client à l'un des segments identifiés.

La solution finale devra être intégrée dans une **application Streamlit** permettant de saisir les caractéristiques d'un nouveau client et d'obtenir automatiquement son segment.

---

# 📌 Contexte

Vous disposez du dataset **Online Retail**, contenant des informations sur les transactions réalisées par une entreprise de commerce en ligne.

Chaque ligne correspond à une ligne de transaction et contient notamment des informations concernant :

- la facture ;
- le produit ;
- la quantité ;
- la date d'achat ;
- le prix unitaire ;
- le client ;
- le pays.

Un même client peut apparaître plusieurs fois dans le dataset.

Vous devrez donc transformer les données transactionnelles afin d'obtenir une **vue au niveau client**, qui pourra ensuite être utilisée pour réaliser la segmentation.

# Partie 2 — Construction des indicateurs RFM

Transformer les données transactionnelles afin d'obtenir **une ligne par client**.

Construire les trois indicateurs RFM :

- **Recency (Récence)** : nombre de jours depuis le dernier achat ;
- **Frequency (Fréquence)** : fréquence des achats du client ;
- **Monetary (Montant)** : montant total dépensé par le client.

La table finale devra permettre d'obtenir, au minimum :

`CustomerID | Recence | Frequence | Montant`

Vous devrez expliquer la signification de ces trois indicateurs et leur intérêt pour la segmentation des clients.

# Partie 4 — Segmentation avec K-Means

## 4.1 Recherche d'un nombre de clusters pertinent

Avant de fixer le nombre final de segments, tester plusieurs valeurs de **K** afin d'étudier le nombre de clusters pertinent pour les données.

Utiliser une ou plusieurs méthodes d'évaluation du clustering, notamment :

- la **méthode du coude (Elbow Method)** ;
- le **coefficient de silhouette (Silhouette Score)**.

Vous devrez :

- tester plusieurs valeurs de K ;
- visualiser les résultats ;
- analyser les résultats obtenus ;
- identifier les valeurs de K qui semblent pertinentes selon les critères utilisés.

Cette analyse permet d'obtenir une indication **algorithmique** sur le nombre de clusters adapté aux données.

---

## 4.2 Prise en compte de la contrainte métier

L'analyse précédente permet d'identifier des valeurs de K pertinentes du point de vue algorithmique.

Cependant, le besoin métier impose de construire **exactement trois classes de consommateurs**.

La valeur finale de K sera donc fixée à :

> **K = 3**

Vous devrez expliquer la différence entre :

- le nombre de clusters suggéré par les méthodes d'évaluation ;
- le nombre de segments finalement retenu en raison de la **contrainte métier**.

Ainsi, le choix final de K résulte de la combinaison de l'analyse des données et du besoin métier.

---

## 4.3 Application du K-Means

Appliquer ensuite l'algorithme **K-Means avec K = 3**.

Vous devrez :

- entraîner le modèle ;
- attribuer un cluster à chaque client ;
- ajouter le cluster obtenu à la table RFM ;
- analyser les caractéristiques des différents clusters.

- le cluster obtenu avec K-Means comme variable cible `y`.

Construire un modèle **Random Forest Classifier** permettant d'apprendre à affecter un client à l'un des trois segments.

Séparer les données en :

- données d'entraînement ;
- données de test.

Veiller à conserver une répartition cohérente des différentes classes lors de cette séparation.

# 📏 Partie 8 — Évaluation du modèle de classification

Évaluer les performances du modèle à l'aide de plusieurs métriques :

- Accuracy ;
- Precision ;
- Recall ;
- F1-score ;
- matrice de confusion.

Utiliser également une **validation croisée à 5 folds** sur les données d'entraînement afin d'évaluer la stabilité des performances du modèle.

Enfin, utiliser le **jeu de test** pour réaliser l'évaluation finale du modèle.

Analyser et interpréter les résultats obtenus.

Une performance élevée indique donc principalement que le Random Forest arrive à reproduire les règles de segmentation produites par le clustering. Elle ne signifie pas que le modèle prédit directement un comportement futur du client.
